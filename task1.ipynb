```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/API_SP.POP.TOTL_DS2_en_csv_v2_33112.csv", skiprows=4)
```

```python
df = df.drop(columns=["Unnamed: 70"])
```

```python
df[["Country Name", "2025"]].head()
```

```python
df["2025"].isnull().sum()
```

```python
df = df.dropna(subset=["2025"])
```

```python
top10 = df.sort_values(by="2025", ascending=False).head(10)

top10
```

```python
df = df[df["Country Code"].str.len() == 3]
```

```python
top10 = df.sort_values(by="2025", ascending=False).head(10)

top10
```

```python
plt.figure(figsize=(10,5))

plt.bar(top10["Country Name"], top10["2025"])

plt.title("Top 10 Most Populated Countries (2025)")
plt.xlabel("Country")
plt.ylabel("Population")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("images/top10_population.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["2025"], bins=20)

plt.title("Population Distribution (2025)")
plt.xlabel("Population")
plt.ylabel("Number of Countries")

plt.tight_layout()

plt.savefig("images/population_histogram.png")

plt.show()
```

# Insights

- India has one of the highest populations in 2025.
- China also has a very large population.
- Most countries have much smaller populations than the top countries.
- Population distribution is uneven across countries.
